```python
#@title --- > Calculadora Financeira

# Conceitos, Estruturas e Funções utilizadas:
  # Importação de biblioteca de calculos matemáticos (math para cálculo de log)
  # Criação de funções personalizadas (def)
  # Uso de funções embutidas e conversão de tipos de dados
  # Uso de métodos de string
  # Uso de estruturas de condicional e loop
  # Uso de f-strings
  # Uso de modos notebook e script

import math  # Importa a biblioteca math para usar funções matemáticas avançadas (como logaritmo)

# Função para calcular o Capital Inicial (Principal)
def calculate_principal(valor_futuro, taxa, anos):
    return valor_futuro / (1 + taxa)**anos  # Fórmula: CI = VF / (1 + i)^t

# Função para calcular a Taxa de Juros
def calculate_taxa(valor_futuro, principal, anos):
    return ((valor_futuro / principal)**(1/anos)) - 1  # Fórmula: i = (VF / CI)^(1/t) - 1

# Função para calcular o Número de Anos
def calculate_anos(valor_futuro, principal, taxa):
    return math.log(valor_futuro / principal) / math.log(1 + taxa)  # Fórmula: t = log(VF / CI) / log(1 + i)

# Função para calcular o Valor Futuro
def calculate_valor_futuro(principal, taxa, anos):
    return principal * (1 + taxa)**anos  # Fórmula: VF = CI * (1 + i)^t

# Função principal para interação com o usuário
def main():
    print("Bem-vindo à calculadora de Juros Compostos!")  # Mensagem de boas-vindas

    while True:
        # Pergunta ao usuário qual cálculo ele deseja fazer
        escolha = input("O que você gostaria de calcular? (vf, ci, t, i): ").lower()

        # Verifica se a escolha está entre as opções válidas
        if escolha in ['vf', 'ci', 't', 'i']:
            if escolha == "vf":  # Calcular Valor Futuro
                principal = float(input("Digite o valor principal (capital inicial): "))
                taxa = float(input("Digite a taxa de juros (em decimal): "))
                anos = float(input("Digite o número de anos: "))
                valor_futuro = calculate_valor_futuro(principal, taxa, anos)
                print(f"O Valor Futuro é: {valor_futuro:.2f}")  # Exibe o resultado formatado

            elif escolha == "ci":  # Calcular Capital Inicial
                valor_futuro = float(input("Digite o valor futuro: "))
                taxa = float(input("Digite a taxa de juros (em decimal): "))
                anos = float(input("Digite o número de anos: "))
                principal = calculate_principal(valor_futuro, taxa, anos)
                print(f"O Capital Inicial é: {principal:.2f}")

            elif escolha == "i":  # Calcular Taxa de Juros
                valor_futuro = float(input("Digite o valor futuro: "))
                principal = float(input("Digite o valor principal (capital inicial): "))
                anos = float(input("Digite o número de anos: "))
                taxa = calculate_taxa(valor_futuro, principal, anos)
                print(f"A Taxa de Juros é: {taxa:.4f}")  # Exibe taxa com 4 casas decimais

            elif escolha == "t":  # Calcular Número de Anos
                valor_futuro = float(input("Digite o valor futuro: "))
                principal = float(input("Digite o valor principal (capital inicial): "))
                taxa = float(input("Digite a taxa de juros (em decimal): "))
                anos = calculate_anos(valor_futuro, principal, taxa)
                print(f"O Número de Anos é: {anos:.2f}")

            # Pergunta ao usuário se deseja continuar ou sair
            continuar = input("Deseja fazer outro cálculo? (s/n): ").lower()
            if continuar != "s":
                print("Se precisar fazer mais cálculos, volte a me executar.")
                break  # Sai do loop e encerra o programa

        else:
            print("Escolha inválida. Por favor, escolha entre 'vf', 'ci', 't' ou 'i'.")

# Verifica se o script está sendo executado diretamente e chama a função main
if __name__ == "__main__":
    main()
```
